## Creating a File management System Project

First, I will create a <b> Tkinter window </b> for my File Setup To endure i get started of all well

```python
import os
import shutil
import tkinter as tk
from tkinter import filedialog, messagebox

# Initialize main window
root = tk.Tk()
root.title("File Management System")
root.geometry("500x400")

# Function to browse for a directory
def browse_directory():
    folder_path = filedialog.askdirectory()
    if folder_path:
        folder_entry.delete(0, tk.END)
        folder_entry.insert(0, folder_path)

# Create a folder entry and browse button
folder_label = tk.Label(root, text="Select Folder:")
folder_label.pack(pady=10)

folder_entry = tk.Entry(root, width=50)
folder_entry.pack(pady=5)

browse_button = tk.Button(root, text="Browse", command=browse_directory)
browse_button.pack(pady=5)

```

Second, is ensuring i Organize all my files by extension

```python
def organize_files():
    folder_path = folder_entry.get()
    if not folder_path:
        messagebox.showerror("Error", "Please select a folder!")
        return

    # Check if the folder exists
    if not os.path.isdir(folder_path):
        messagebox.showerror("Error", "Folder does not exist!")
        return

    # Loop through files in the directory
    for file_name in os.listdir(folder_path):
        file_path = os.path.join(folder_path, file_name)
        
        # Skip directories
        if os.path.isdir(file_path):
            continue

        # Get the file extension
        file_extension = file_name.split('.')[-1]
        new_folder = os.path.join(folder_path, file_extension)
        
        # Create folder if it doesn't exist
        if not os.path.exists(new_folder):
            os.makedirs(new_folder)
        
        # Move file to the new folder
        shutil.move(file_path, os.path.join(new_folder, file_name))

    messagebox.showinfo("Success", "Files organized by extension!")

# Add a button to trigger the organize function
organize_button = tk.Button(root, text="Organize Files", command=organize_files)
organize_button.pack(pady=10)
```

Fourth, is the File Deleter.
This is in case we need to delete a folder name, we can perform it without problems

```python
def delete_files():
    folder_path = folder_entry.get()
    if not folder_path:
        messagebox.showerror("Error", "Please select a folder!")
        return

    # Confirm deletion
    response = messagebox.askyesno("Confirmation", "Are you sure you want to delete all files?")
    if not response:
        return

    # Loop through files in the directory
    for file_name in os.listdir(folder_path):
        file_path = os.path.join(folder_path, file_name)

        # Skip directories
        if os.path.isdir(file_path):
            continue

        # Delete the file
        os.remove(file_path)

    messagebox.showinfo("Success", "Files deleted successfully!")

# Add a button to trigger the delete function
delete_button = tk.Button(root, text="Delete Files", command=delete_files)
delete_button.pack(pady=10)
```

Fifth, Is moving files from another Directory

```python
def move_files():
    folder_path = folder_entry.get()
    if not folder_path:
        messagebox.showerror("Error", "Please select a folder!")
        return

    # Browse for destination folder
    dest_folder = filedialog.askdirectory()
    if not dest_folder:
        return

    # Loop through files in the directory
    for file_name in os.listdir(folder_path):
        file_path = os.path.join(folder_path, file_name)

        # Skip directories
        if os.path.isdir(file_path):
            continue

        # Move file to the new folder
        shutil.move(file_path, os.path.join(dest_folder, file_name))

    messagebox.showinfo("Success", "Files moved successfully!")

# Add a button to trigger the move function
move_button = tk.Button(root, text="Move Files", command=move_files)
move_button.pack(pady=10)
```
